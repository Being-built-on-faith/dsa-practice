```python
boxes = ['hi','hello','some_other_item','some other item2','bob']
target_item = 'bob'
def linear_search(boxes, target_item):
    for _ in range(len(boxes)):
        if boxes[_]==target_item:
            return 1
    print(f"target item i.e. {target_item} is not found.")
    return -1
linear_search(boxes,target_item)
```

```python
boxes = [
    "Apple",
    "Banana",
    "Cat",
    "Dog",
    "Egg"
]

def binary_search(boxes, var="Dog"):
    left = 0 
    right = len(boxes) -1
    while left<=right:
        mid = (left + right)//2
        guess = boxes[mid]
        if guess == var:
            return mid
        if guess<var:
            left = mid+1
        elif guess>var:
            right = mid-1
        else:
            print(f"not found at index{mid}")
    return -1

bob = binary_search(boxes,var="Dogg")

if bob != -1:
    print(f"the item is found at index{bob}")       
else:
    print("Didnt found the match")     
```
