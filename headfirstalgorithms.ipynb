📘 Jupyter Notebook Shortcuts (Structured Guide)

🧠 Modes Overview
    Command Mode (Blue Border): For notebook-level actions (add/delete/navigate cells) → press Esc
    Edit Mode (Green Border): For typing/editing inside a cell → press Enter

1️⃣ Adding & Changing Cells (Command Mode)
    Action                         Shortcut            Memory Hook
    Add cell below                 B                   Below → B
    Add cell above                 A                   Above → A
    Convert to Markdown            M                   Markdown → M
    Convert to Code                Y                   (Think: Python → Y)
    Delete cell                    D → D               Double tap = destructive
    Undo delete                    Z                   Undo → Z (Ctrl+Z analogy)

2️⃣ Running & Navigating Cells
    Action                         Shortcut            Behavior
    Run + move down                Shift + Enter       Most used (default flow)
    Run + stay                     Ctrl + Enter        Debug / re-run same cell
    Run + insert new below         Alt + Enter         Fast notebook building
    Move between cells             ↑ / ↓               Navigation
    Multi-select cells             Shift + ↑ / ↓       Batch operations

3️⃣ Editing Cells (Edit Mode)
    (Press Enter to enter Edit Mode)

    Action                         Shortcut            Insight
    Split cell                     Ctrl + Shift + -    Cursor defines split point
    Merge cells                    Shift + M           Works on selected cells
    Toggle comment                 Ctrl + /            Fast debugging

4️⃣ Advanced Utility & Power Moves
    Action                         Shortcut            Why it matters
    Command Palette                Ctrl + Shift + P    Search anything
    Show all shortcuts             H (Command Mode)    Discoverability
    Save notebook                  S or Ctrl + S       Avoid losing work

🧠 Mental Model (Memory Compression)
    Think in 3 layers:
        Structure layer (Command Mode) → A, B, D, Z
        Execution layer → Shift+Enter, Ctrl+Enter
        Editing layer (Edit Mode) → Split, Merge, Comment

⚡ Plain-English Recap
    Jupyter is basically a two-mode system: one mode controls the notebook (add/delete/move cells), the other edits code. Most efficiency comes from mastering just 5 keys: A, B, Shift+Enter, D+D, Enter. Everything else is optimization on top.

⚠️ Pitfalls
    Staying in wrong mode → shortcuts “don’t work”
    Accidental D+D → deletes cell (use Z)
    Overusing mouse → kills speed (keyboard is 10x faster)

```python
boxes = ["hi", "hola", "amigo"]

keyword = "hii"
def linear_search(boxes, keyword):
    for var in range(len(boxes)):
        if boxes[var] == keyword:
            return 1
        else:
            print(f"Keyword not in box {var}")
    return -1 
    
bob = linear_search(boxes, keyword)
print(bob)
        
```
