```python
users = {'Hans':'active', 'bob':'inactive', 'sam':'active'}
for user, status in users.copy().items():
    if status == "inactive":
        del users[user]
    print(users.copy())
```

```python
active_users = {}
for user, status in users.items():
    print(user,status)
    print("============")
    print("users[user]=",users[user])
    if status == 'active':
        active_users[user]= status
    print(active_users)
```

```python
for n in range(2, 10):
    for x in range(2,n):
        print("n is :",n)
        print("x is :",x)
        bob = n % x
        print("bob",bob)
        if bob == 0:
            print(f"{n} equals {x} * {n//x}")
            break
    print("=========")
```
